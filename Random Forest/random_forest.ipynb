## Preprocessing

```python
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

target = "loanDefault"

x = df.drop(columns=[target, "loanId"])
y = df[target]

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)

print("Train:", x_train.shape, "Test :", x_test.shape)
print("Rasio gagal bayar train", round(y_test.mean(), 3),"| test", round(y_test.mean(),3))
```

```python
fitur_num = x.select_dtypes(include="number").columns.to_list()
fitur_kat = x.select_dtypes(include="number").columns.to_list()
print("Numerik:", fitur_num)
print("Katergori:", fitur_kat)

prep_pohon = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), fitur_num),
    ("kat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
                      ("ohe", OneHotEncoder(handle_unknown="ignore"))]), fitur_kat),
])
```

```python
model_rf = Pipeline([("prep", prep_pohon),
                     ("clf", RandomForestClassifier(n_estimators=300, max_depth=8, min_samples_leaf=10,
                                                    class_weight="balanced", n_jobs=-1, random_state=RANDOM_STATE))])
model_rf.fit(x_train, y_train)
```

## Evaluasi Model

```python
def evaluasi(nama, model, X_test, y_test, threshold=0.5):
    proba = model.predict_proba(X_test)[:, 1]
    pred = (proba >= threshold).astype(int)
    return {"model": nama,
            "accuracy": accuracy_score(y_test, pred),
            "precision": precision_score(y_test, pred, zero_division=0),
            "recall": recall_score(y_test, pred),
            "f1": f1_score(y_test, pred),
            "roc_auc": roc_auc_score(y_test, proba)}

hasil = [evaluasi("Random Forest", model_rf, x_test, y_test)]
pd.DataFrame(hasil).round(3)
```

```python
threshold = 0.5
proba = model_rf.predict_proba(x_test)[:, 1]
pred = (proba >= threshold).astype(int)

report = classification_report(y_test, pred, target_names=['Lancar (0)', 'Gagal Bayar (1)'])
print(report)
ConfusionMatrixDisplay.from_predictions(y_test, pred, display_labels=["Lancar", "Gagal bayar"], cmap="Blues")
plt.title("Confusion Matrix - Random Forest"); plt.show()
```

```python
imp = pd.Series(model_rf.named_steps["clf"].feature_importances_,
                index=model_rf.named_steps["prep"].get_feature_names_out()).sort_values()
imp.tail(10).plot(kind="barh", figsize=(7, 4), color="seagreen")
plt.title("10 fitur terpenting (Random Forest)"); plt.tight_layout(); plt.show()
```
